Problem 2 — Conditional Filtering

Find rows where:

• one feature is above its column median
• another feature is below its column mean

Create a dataset containing only those rows.

Then sort those rows by the difference between those two features.

```python
beast = df.to_numpy()
beast
```

```python
col_mean = np.mean(beast,axis=0)
col_median = np.median(beast,axis=0)
meanMask = beast < col_mean
meadianMask = beast > col_median
```

```python
#np.argwhere(meanMask==True)
```

```python
mi = pd.DataFrame(np.where(meanMask==True)).T
mi
```

```python
mi.drop_duplicates(subset=0,keep='first',inplace=True)
mi = mi.to_numpy()
mi
```

```python
mdi = pd.DataFrame(np.where(meadianMask==True)).T
mdi.drop_duplicates(subset=0,keep='first',inplace=True)
mdi
```

```python
c1 = beast[mi[:,0],mi[:,1]]
c1
```

```python
c2 = beast[mdi.iloc[:,0],mdi.iloc[:,1]]
c2
```

```python
c2 = np.append(c2,values=0)
c2
```

```python
answer = pd.DataFrame({'BelowMean':c1.T,'AboveMedian':c2.T})
answer
```

```python
answer.iloc[(answer.iloc[:,0]-answer.iloc[:,1]).argsort()]
```

```python
answer.agg(['min','max'])
```
