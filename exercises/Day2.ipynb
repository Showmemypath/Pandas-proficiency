```python
n0 = {"Monday":[2,3,4,5],"Tuesday":[6,7,8,9]}
n1 = [[2,3,4,5],[6,7,8,9]]
n2 = {"Wednusday":{"ALA":30,"CA":60},"Thursday":{"EWASTE":4,"RM":2},"Monday":[2,3,4,5],"Tuesday":[6,7,8,9]}
n4 = {"Wednusday":{"ALA":30,"CA":60},"Thursday":{"EWASTE":4,"RM":2}}
n3 = {5:"Friday",6:"Saturday",7:"Monday",8:"Tuesday"}

print(pd.DataFrame(n1),'\n')

pd.DataFrame(n0,n2) # n2 is considered row index of n1 

print(pd.DataFrame(n1,n4,n3),'\n') # n3 is considered column index of n1

# print(pd.DataFrame(n2),'\n') Will raise error
print(pd.DataFrame(n3,index=[1,2,3,4]),'\n')
```

```python
#         keys = col name : value = dtype
dtype = {'int64':'int64',
          'float64':'float64',
          'bool':'bool',
          'category':'category'}
data = pd.DataFrame({'int64':[1,2],
        'float64':[3.9,4],
        'bool':[True,False],
        'category':('a','b')})
data = data.astype(dtype)
display(data)
data.info()
```
