```python
import pandas as pd
import numpy as np
```

Series at default follow label base indexing
this 

`s = pd.Series({'apple': 5, 'banana': 3, 'orange': 8})`
`s["apple"]` would work not , `s[1]` unless 1 is index label

`s.loc[]` -> Label based indexing
`s.iloc[]`-> position/index based indexing

Series allow boolean indexing 
`s[s>3]`


```python
s = pd.Series({'apple': 5, 'banana': 3, 'orange': 8})
print(s[s>3])
s.iloc[0]
```
